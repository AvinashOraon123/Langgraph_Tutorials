```python
from langgraph.graph import StateGraph , START , END
from langchain_google_genai import GoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
model = GoogleGenerativeAI(model="gemini-3.6-flash", temperature=0.2)
```

```python
# create a state

class LLMState(TypedDict):
    prompt: str
    response: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    #extract the prompt from the state
    question = state['prompt']

    #form a prompt for the model
    state['prompt'] = f"Answer the following question: {question}"

    #a=sk the model to answer the question
    response = model.invoke(state['prompt'])

    #store the response in the state
    state['response'] = response
```

```python
#create our graph

graph = StateGraph(LLMState)

#ADD NODES

graph.add_node('llm_qa',llm_qa)

#add edges to the graph
graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)

workflow = graph.compile()
```

```python
initial_state = LLMState(prompt="What is the capital of France?", response="")

final_state = workflow.invoke(initial_state)
print(final_state['response'])
```
